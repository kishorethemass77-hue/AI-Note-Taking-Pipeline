```python
import re
from collections import Counter

def ai_note_summarizer(text, summary_ratio=0.4):
    text = re.sub(r"\s+", " ", text).strip()

    if not text:
        return "Please enter some text."

    sentences = re.split(r'(?<=[.!?])\s+', text)
    sentences = [s.strip() for s in sentences if s.strip()]

    if len(sentences) <= 2:
        return " ".join(sentences)

    stop_words = {
        "the", "is", "a", "an", "and", "or", "to", "of",
        "in", "on", "for", "with", "are", "was", "by",
        "it", "this", "that", "as", "be", "can", "has"
    }

    words = re.findall(r"\b[a-zA-Z]+\b", text.lower())
    frequencies = Counter(
        word for word in words if word not in stop_words
    )

    scored_sentences = []

    for index, sentence in enumerate(sentences):
        sentence_words = re.findall(
            r"\b[a-zA-Z]+\b", sentence.lower()
        )

        score = sum(
            frequencies[word]
            for word in sentence_words
        )

        if sentence_words:
            score /= len(sentence_words)

        scored_sentences.append((score, index, sentence))

    count = max(1, round(len(sentences) * summary_ratio))

    best = sorted(
        scored_sentences,
        key=lambda item: item[0],
        reverse=True
    )[:count]

    best.sort(key=lambda item: item[1])

    return " ".join(item[2] for item in best)


print("AI NOTE-TAKING PIPELINE")
print("-" * 30)

text = input("Enter your notes:\n")

summary = ai_note_summarizer(text)

print("\nGENERATED SUMMARY:")
print(summary)
```
